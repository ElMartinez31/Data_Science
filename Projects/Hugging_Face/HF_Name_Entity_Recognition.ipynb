```python
import kagglehub
import pandas as pd
import os

# Download latest version
path = kagglehub.dataset_download("modhiibrahimalmannaa/1000-children-books-on-amazom")

print("Path to dataset files:", path)

os.listdir(path)

df = pd.read_csv(path + "/" + "books_data.csv")
df
```

```python
#####################
# HF distillbert NER
#####################

from transformers import AutoTokenizer, AutoModelForTokenClassification
from transformers import pipeline

#tokenizer = AutoTokenizer.from_pretrained("dslim/distilbert-NER")
#model = AutoModelForTokenClassification.from_pretrained("dslim/distilbert-NER")

model = "dbmdz/bert-large-cased-finetuned-conll03-english"

nlp = pipeline("ner", model=model, device=-1, aggregation_strategy="simple")
example = "My name is Wolfgang and I live in Berlin"

ner_results = nlp(example)
print(ner_results)
```

```python
# Extract names

ner_results[0:2]
```

```python
nlp(df.Description[0])
```

```python
[entity["word"] for entity in nlp(df.Description[0]) if entity["entity_group"] == "PER"]
# if entity["entity_group"] == "PER" to filter on people only
```

```python

res = df.Description[:20].dropna().apply(lambda x : [entity["word"] for entity in nlp(x) if entity["entity_group"] == "PER"] )
```

```python
type(res) # series
```

```python
names = list(set(res.explode().tolist()))
names
# we still have subword
```

```python
filtered_names = [entity for entity in names
                  if isinstance(entity, str) and entity.strip() != '' and '#' not in entity]
filtered_names
```
