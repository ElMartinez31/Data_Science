```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
from transformers import pipeline
```

```python
import kagglehub
import pandas as pd
import os

# Download latest version
path = kagglehub.dataset_download("modhiibrahimalmannaa/1000-children-books-on-amazom")

print("Path to dataset files:", path)

os.listdir(path)

df = pd.read_csv(path + "/" + "books_data.csv")
df
```

```python
model = "facebook/bart-large-mnli" # encoder decoder
classifier = pipeline("zero-shot-classification",
                      model = model,
                      device = "cpu")
```

```python
#example

text1 = "When life gives you lemons, make lemonade!"
text2 = "In god we trust"
text3 = "Hambuger is made of steak, brad and cheese"

texts = [text1, text2, text3]
```

```python
# example

for text in texts:
  res = classifier(text, ["quote", "food & drinks", "religion", "animals"])
  print(res)
```

```python
labels = ["quote", "food & drinks", "religion", "animals", "Adventures", "Litterature"]

df.loc[:5,"Category"] = df.Description[:5].apply(lambda row : classifier(row, labels)["labels"][0]) # get the main detected label + assign it to df
```
