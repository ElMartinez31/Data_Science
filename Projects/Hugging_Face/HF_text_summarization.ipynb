```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
from transformers import pipeline
import kagglehub
import pandas as pd
import os

# Download latest version
path = kagglehub.dataset_download("modhiibrahimalmannaa/1000-children-books-on-amazom")

print("Path to dataset files:", path)

os.listdir(path)

df = pd.read_csv(path + "/" + "books_data.csv")
df
```

```python
df["Description"] = df["Description"].str.replace('\n', ' ', regex=False)
df
```

```python
model_summary = "facebook/bart-large-cnn" # encoder decoder
summarizer = pipeline("summarization",
                      model = model_summary,
                      device = "cpu")

model_sentiment = ""
```

```python
summarizer(df.Description[0])
```

```python
summarizer(df.Description[0])[0]["summary_text"]
```

```python
df.loc[:5, "Summary"] = df.Description[:5].apply(lambda row : summarizer(row, max_length = 70)[0]["summary_text"])
```
