1️⃣ Comment fonctionne Streamlit en interne

Streamlit transforme un script Python normal (.py) en application web interactive.

🔹 Étapes internes :

```python
import streamlit as st
st.title("Hello Streamlit")
x = st.slider("Choisis un nombre", 0, 100, 50)
st.write("Tu as choisi :", x)
```

```python
# pour executer:
streamlit run mon_app.py # il faut bien que ça soit un .py et pas notebook (on peut convertir facielemnt)
```

Streamlit :

démarre un petit serveur web local (Flask-like)

lit ton code ligne par ligne

affiche les widgets et les résultats dans une page web interactive

À chaque fois que tu interagis avec un élément (slider, selectbox, bouton, etc.) :

le navigateur envoie la nouvelle valeur au serveur Streamlit,

le script entier est réexécuté,

Streamlit conserve l’état des widgets via st.session_state

et met à jour uniquement ce qui a changé dans la page.

🌀 En clair :

Streamlit “rejoue ton script” à chaque interaction, mais de façon intelligente et réactive.

2️⃣ Comment Streamlit “parle” à ton code Python

Tout ce que tu écris (st.title, st.write, st.plotly_chart, etc.) ne fait pas d’affichage console,
mais envoie un objet au “frontend” Streamlit.

Ce frontend (une app React/TypeScript) s’occupe de l’afficher joliment dans le navigateur.

Exemple :

```python
import streamlit as st
import pandas as pd

df = pd.DataFrame({
    "Nom": ["Alice", "Bob", "Charlie"],
    "Score": [85, 90, 95]
})

st.title("Scores des étudiants")
st.table(df)

#Quand tu lances l’app :

#Python → calcule le DataFrame

#Streamlit → sérialise le DataFrame en JSON

#Le navigateur → le rend sous forme de tableau HTML interactif.
```

3️⃣ L’intégration avec Plotly (et autres bibliothèques de visualisation)

Streamlit ne dessine pas les graphes lui-même.
Il sait afficher les objets graphiques produits par d’autres libs.

🔹 Exemple avec Plotly

Plotly génère une figure interactive (objet Python) :

```python
import plotly.express as px
fig = px.scatter(df, x="capacity", y="numdoc")


```

```python
#Pour l’afficher, Streamlit la “transmet” à son front-end :*

st.plotly_chart(fig)

# ➡️ Streamlit détecte que c’est un objet Plotly,
# ➡️ le convertit en JSON,
# ➡️ et l’affiche via la librairie Plotly.js intégrée dans le navigateur.

# C’est donc une intégration naturelle et transparente :

# tu n’as rien à “convertir” toi-même,

# tu gardes toute l’interactivité native de Plotly.
```
